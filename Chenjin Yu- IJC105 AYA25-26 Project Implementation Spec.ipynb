````python
def getNumericColumns(data_file):
  #connect to sqlite database
  conn = sqlite3.connect(data_file)
  #create cursor object for executing queries
  cursor = conn.cursor()

  # #store a list of numeric columns in the variable
  numeric_cols = ['STOCK_LEVEL', 'QUANTITY', 'PRICE']
  #initialise the dictionary to store values of each column
  numeric_dict = {}
  for col in numeric_cols:
    numeric_dict[col] = []

  #select numeric columns from product and order table
  cursor.execute("SELECT QUANTITY FROM ORDERS")
  #loop over rows, add values to dictionary where key is QUANTITY column
  for row in cursor.fetchall():
    numeric_dict['QUANTITY'].append(row[0])

  cursor.execute("SELECT STOCK_LEVEL, PRICE FROM PRODUCTS")
  #loop over rows, add values to dictionary where keys are STOCK_LEVEL, PRICE columns
  for row in cursor.fetchall():
      numeric_dict['STOCK_LEVEL'].append(row[0])
      numeric_dict['PRICE'].append(row[1])
  #close connection to database
  conn.close()
  return numeric_dict

numeric_columns = getNumericColumns("clothes_retailer.db")
print(numeric_columns)
````

````python
def getCategoricalColumns(data_file):
  #connect to sqlite database
  conn = sqlite3.connect(data_file)
  #create cursor object for executing queries
  cursor = conn.cursor()

  #initialise the dictionary to store values of categorical column
  cate_dict = {}
  cate_dict['CATEGORY'] = []

  cursor.execute("SELECT CATEGORY FROM PRODUCTS")
  #loop over the rows to add values to the dictionary
  for row in cursor.fetchall():
    cate_dict['CATEGORY'].append(row[0])

  conn.close()
  return cate_dict

cate_columns = getCategoricalColumns("clothes_retailer.db")
print(cate_columns)
````

````python
def getFreeTextColumns(data_file):
  #connect to sqlite database
  conn = sqlite3.connect(data_file)
  #create cursor object for executing queries
  cursor = conn.cursor()

  #store a list of free text columns in the variable
  text_cols = ['FORENAME', 'SURNAME', 'PHONE_NUMBER', 'POSTCODE', 'PRODUCT_NAME', 'ORDER_DATE']

  #initialise the dictionary to store values of free text columns
  text_dict = {}
  for col in text_cols:
    text_dict[col] = []

  cursor.execute("SELECT * FROM CUSTOMERS")
  #loop over the rows to add values to the dictionary
  for row in cursor.fetchall():
    text_dict['FORENAME'].append(row[0])
    text_dict['SURNAME'].append(row[1])
    text_dict['PHONE_NUMBER'].append(row[2])
    text_dict['POSTCODE'].append(row[3])

  cursor.execute("SELECT PRODUCT_NAME FROM PRODUCTS")
  for row in cursor.fetchall():
    text_dict['PRODUCT_NAME'].append(row[0])

  cursor.execute("SELECT ORDER_DATE FROM ORDERS")
  for row in cursor.fetchall():
    text_dict['ORDER_DATE'].append(row[0])

  conn.close()
  return text_dict

text_columns = getFreeTextColumns("clothes_retailer.db")
print(text_columns)
````

````python
def getAllRows(data_file):
  conn = sqlite3.connect(data_file)
  cursor = conn.cursor()

  #initialise lists to store each row where each row will be a dictionary
  customer_table = []
  product_table = []
  order_table = []

  cursor.execute("SELECT * FROM CUSTOMERS")
  #loop over the rows to assign values to the dictionary
  for row in cursor.fetchall():
    row_dict = {
        "PHONE_NUMBER": row[0],
        "FORENAME": row[1],
        "SURNAME": row[2],
        "POSTCODE": row[3]
    }
    #add dictionaries/rows to the customer_table list
    customer_table.append(row_dict)

  cursor.execute("SELECT * FROM PRODUCTS")
  #loop over the rows to assign values to the dictionary
  for row in cursor.fetchall():
    row_dict = {
        "PRODUCT_NAME": row[0],
        "CATEGORY": row[1],
        "STOCK_LEVEL": row[2],
        "PRICE": row[3]
    }
    #add dictionaries/rows to the product_table list
    product_table.append(row_dict)

  cursor.execute("SELECT * FROM ORDERS")
  #loop over the rows to add values to the dictionary
  for row in cursor.fetchall():
    row_dict = {
        "ORDER_DATE": row[0],
        "PHONE_NUMBER": row[1],
        "PRODUCT_NAME": row[2],
        "QUANTITY": row[3]
    }
    #add dictionaries/rows to the order_table list
    order_table.append(row_dict)

  #initialise a dictionary, where key is each table name, and value is a list of rows/dictionaries
  table = {}
  table["CUSTOMERS TABLE"] = customer_table
  table["PRODUCTS TABLE"] = product_table
  table["ORDERS TABLE"] = order_table

  return table
  conn.close()

getAllRows("clothes_retailer.db")
````

````python
def getOrderDate(prompt):
  #initialise date format
  date_format = "%d/%m/%Y"

  #while loop will keep asking for user input until the function returns the order_date variable
  while True:
    try:
      #convert string date to datetime object and then change date format
      date_obj = datetime.strptime(input(prompt), date_format)
      order_date = date_obj.strftime(date_format)
      return order_date

    #raise error if format doesn't match
    except ValueError:
      print("Please enter a valid order date using the format: DD/MM/YYYY")

order_date = getOrderDate("Enter order date: ")
````

````python
def getStockLevel(product_name):
  cursor = conn.cursor()
  #retrieve original stock level
  cursor.execute("SELECT STOCK_LEVEL FROM PRODUCTS WHERE PRODUCT_NAME = ?", (product_name,))
  for row in cursor:
    return row[0]
````

````python
def updateStock(quantity, product_name):
  pre_stockLevel = getStockLevel(product_name)
  print(f"The stock level of {product_name} before the new order was {pre_stockLevel}.")

  #update stock level after a new order is inserted
  print("Updating Stock Level...")
  cursor = conn.cursor()
  query = "UPDATE PRODUCTS SET STOCK_LEVEL = STOCK_LEVEL - ? WHERE PRODUCT_NAME = ?"
  cursor.execute(query, (quantity, product_name))
  conn.commit()

  #check if the stock level is updated
  new_stockLevel = getStockLevel(product_name)
  print(f"The current stock level of {product_name} is {new_stockLevel}.")

updateStock(int(quantity), str(product_name.value))
````

````python
class BaseSummaryReport:
  def printCustomContent(self):
    return NotImplemented

  def __init__(self, data_file):
    self.data_file = data_file
    self.printReportHeader()
    self.printCustomContent()
    self.printReportFooter()

  def printReportHeader(self):
    print('-'*80)
    print('>>> SUMMARY REPORT <<<')
    print()
    print('Using data file:', self.data_file)
    print()

    print()
    print('== TABLE METADATA== ')
    print()

    rows = getAllRows(self.data_file)
    for (tbl_name, tbl) in sorted(rows.items()):
      print('Table: {0}'.format(tbl_name))
      print('No. rows: {0:,}'.format(len(tbl)))
      print()
      all_cols = set(tbl[0].keys())

    print()
    print('== COLUMN METADATA ==')
    print()

    numeric = getNumericColumns(self.data_file)
    numeric = set(numeric.keys())
    print('Numeric columns:', ', '.join(sorted(numeric)))

    categorical = getCategoricalColumns(self.data_file)
    categorical = set(categorical.keys())
    print('Categorical columns:', ', '.join(sorted(categorical)))

    free_text = getFreeTextColumns(self.data_file)
    free_text = set(free_text.keys())
    print('Free-text columns:', ', '.join(sorted(free_text)))

    other = all_cols - numeric - categorical - free_text
    print('Other columns:', ', '.join(sorted(other)))

    print()
    print()

  def printReportFooter(self):
    print('>>> END SUMMARY REPORT <<<')
    print('-'*80)
````

````python
class SummaryReport(BaseSummaryReport):
  def getColumnStatistics(self, values):
    #get minimum and maxmum values using min() and max() functions
    minimum_value = min(values)
    maximum_value = max(values)

    #initialize sum for calculating mean
    sum = 0
    #create a list for calculating standard deviation
    value_list = []

    for value in values:
      #calculate sum of values
      sum += value
      #add value to the list
      value_list.append(value)
    #calculate mean value
    avg = sum/len(values)
    #calculate standard deviation
    std = np.std(value_list)

    return minimum_value, maximum_value, avg, std

  def getCategoricalInfo(self, values):
    #create a dictionary to store the frequency of each category
    count = {}
    for value in values:
      #if the category is not in the dictionary, add a new count of 0
      if not value in count:
        count[value] = 0
      #increment the count by 1
      count[value] = count[value] + 1

    return count

  def getFreeTextInfo(self, values):
    #initialise shortest and longest text as the first value in the dictionary values
    shortest_text = values[0]
    longest_text = values[0]

    #loop over the index to access each value in the values list
    for i in range(len(values)):
      #if current text length is shorter than the current shortest text,
      #assign current text as shortest text
      if len(values[i]) < len(shortest_text):
        shortest_text = values[i]
      else:
        shortest_text = shortest_text
      #if current text length is longer than the current longest text,
      #assign current text as longest text
      if len(values[i]) > len(longest_text):
        longest_text = values[i]
      else:
        longest_text = longest_text

    return shortest_text, longest_text

  def getPopularCate(self):
    #retrieve most popular category by calculating total quantity sold
    cate_quantity_df = pd.read_sql("SELECT CATEGORY, SUM(QUANTITY) AS TOTAL_QUANTITY FROM PRODUCTS, ORDERS WHERE ORDERS.PRODUCT_NAME = PRODUCTS.PRODUCT_NAME GROUP BY CATEGORY ORDER BY TOTAL_QUANTITY DESC", conn)
    #create a bar plot where x is category and y is total quantity sold
    plt.bar(cate_quantity_df["CATEGORY"], cate_quantity_df["TOTAL_QUANTITY"], color = 'green', width= 0.6)
    plt.xlabel("Product Category")
    plt.ylabel("Total Quantity Sold")
    plt.title("The Most Popular Product Category")
    plt.show()

  def getBestSellingProduct(self):
    #retrieve besting selling product by multiplying price by quantity
    total_revenue_by_product = cursor.execute("SELECT PRODUCTS.PRODUCT_NAME, SUM(PRICE*QUANTITY) AS TOTAL FROM PRODUCTS, ORDERS WHERE ORDERS.PRODUCT_NAME = PRODUCTS.PRODUCT_NAME GROUP BY PRODUCTS.PRODUCT_NAME ORDER BY TOTAL DESC LIMIT 5")
    return total_revenue_by_product.fetchall()

  def getLowStock(self):
    #retrieve products when the stock left is less than or equal to 5
    stock_by_product = cursor.execute("SELECT PRODUCT_NAME, STOCK_LEVEL FROM PRODUCTS WHERE NOT STOCK_LEVEL > 5")
    return stock_by_product.fetchall()

  def printCustomContent(self):
    print("-"*80)

    #num_col_dict is a dictionary where numeric columns are keys, and column values are values
    num_col_dict = getNumericColumns(self.data_file)
    for key in num_col_dict:
      min, max, avg, std = self.getColumnStatistics(num_col_dict[key])
      print("Numerical Column Name: ", key)
      print("Mean: ", avg)
      print("Minimum: ", min)
      print("Maximum: ", max)
      print("Standard Deviation: ", std)
      print()

    print("-"*80)

    #cate_col_dict is a dictionary where categorical columns are keys, and column values are values
    cate_col_dict = getCategoricalColumns(self.data_file)
    for key in cate_col_dict:
      print("Categorical Column Name: ", key)
      print("Below is the frequency of each category:")
      print(self.getCategoricalInfo(cate_col_dict[key]))
      print()

    print("-"*80)

    #text_col_dict is a dictionary where free text columns are keys, and column values are values
    text_col_dict = getFreeTextColumns(self.data_file)
    for key in text_col_dict:
      shortest_str, longest_str = self.getFreeTextInfo(text_col_dict[key])
      print("Free Text Column Name: ", key)
      print("Shortest String: ", shortest_str)
      print("Longest String: ", longest_str)
      print()

    print("-"*80)

    #print bar chart
    self.getPopularCate()

    print()
    #print total revenue for each product
    revenue_by_product = self.getBestSellingProduct()
    print(f"Top Revenue Product: {revenue_by_product}")

    print()
    #print products with stock level less than or equal to 5
    low_stock_pro = self.getLowStock()
    print(f"Products with low stock level: {low_stock_pro}.")
````
